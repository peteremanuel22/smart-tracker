```python
FIELDS = ("date", "category", "amount_egp", "payment_method", "quantity")
# TODO: ﻋﺪل اﻷﺳﻤﺎء ﻟﺤﻘﻞ ﻛﺎرت ﻣﺸﺮوﻋﻚ أﻧﺖ ﻟﻮ ﻣﺨﺘﻠﻔﯿﻦ
categories = set(r["category"] for r in records)   # ﻛﻮد ﺟﺎھﺰ: ﻗﯿﻢ اﻟ
```
