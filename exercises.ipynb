```python
grades = {"math": 88, "english": 76}
grades["math"] = 92         # ﺗﻌﺪﯾﻞ: ﻧﻔﺲ اﻟﻤﻔﺘﺎح، ﻗﯿﻤﺔ ﺟﺪﯾﺪة #
grades["physics"] = 79
del grades["english"]       # ﻣﻔﺘﺎح ﺟﺪﯾﺪ = إﺿﺎﻓﺔ
#ﺣﺬف #

best = max(grades.values())  # أﻋﻠﻰ درﺟﺔ — max() داﻟﺔ ﺟﺎھﺰة ﻓﻲ ﺑﺎﯾﺜﻮن
print(grades, best)
```

```python
assert grades == {"math": 92, "physics": 79} # "math اﺗﻌﺪﻟﺖ، english "اﺗﺤﺬﻓت
assert best == 92 # "best = أﻋﻠﻰ ﻗﯿﻤﺔ ﻓﻲ grades.values()"
print("A3: ﻣﻤﺘﺎز")
```
